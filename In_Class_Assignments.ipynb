```python
class Greeting:
  def __init__(self,first,last):
    self.first = first
    self.last = last

  def display(self, formal=False):
    if formal:
     print(f'Hello {self.first} {self.last}!')
    else:
      print(f'Hi, {self.first}')

  def __str__(self):
   return str(self.first) + ", " + str(self.last)
   return "{}, {}".format(self.first, self.last)
   return f'{self.first}, {self.last}'

greeting = Greeting('Donald', 'Becker')
other_greeting = Greeting('Bob', 'Dylan')
greeting.display()
print(greeting)
print(other_greeting)
```

```python
import math

class Point:
  def __init__(self,x,y):
    self.x = x
    self.y = y

  def distance(self, other):
     return math.sqrt((self.x-other.x)**2 + (self.y-other.y)**2)

  def __str__(self):
    return f'({self.x}, {self.y})'

class Rectangle:
  def __init__(self):
    self.corner_1 = Point(0,0)
    self.corner_2 = Point(0,2)
    self.corner_3 = Point(2,0)
    self.corner_4 = Point(2,2)
    self.width = self.calculate_width()
    self.length = self.calculate_length()

  def calculate_width(self):
   return self.corner_1.distance(self.corner_2)

  def calculate_length(self):
   return self.corner_1.distance(self.corner_3)

  def area(self):
   return f'Area = {self.width * self.length} sq.ft'

square = Rectangle()
square.area()
```
